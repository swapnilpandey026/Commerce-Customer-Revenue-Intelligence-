CUSTOMER LEVEL SUMMARY

  "actual customer intellligence"

```python
customer_summary = df_clean.groupby('CustomerID').agg(
    TotalRevenue=('Revenue', 'sum'),
    TotalOrders=('InvoiceNo', 'nunique'),
    TotalUnits=('Quantity', 'sum'),
    AverageOrderValue=('Revenue', 'mean'),
    FirstPurchase=('InvoiceDate', 'min'),
    LastPurchase=('InvoiceDate', 'max')
).reset_index()

customer_summary.head()
```

CUSTOMER SEGMENTATION

```python
def customer_segment(revenue):
    if revenue >= customer_summary['TotalRevenue'].quantile(0.90):
        return 'High Value'
    elif revenue >= customer_summary['TotalRevenue'].quantile(0.50):
        return 'Medium Value'
    else:
        return 'Low Value'

customer_summary['CustomerSegment'] = (
    customer_summary['TotalRevenue']
    .apply(customer_segment)
)

customer_summary['CustomerSegment'].value_counts()
```

ANALYZE REVENUE BY SEGMENTATION

```python
segment_analysis = (
    customer_summary
    .groupby('CustomerSegment')
    .agg(
        Customers=('CustomerID', 'count'),
        Revenue=('TotalRevenue', 'sum'),
        AvgRevenue=('TotalRevenue', 'mean')
    )
    .sort_values('Revenue', ascending=False)
)

segment_analysis
```

RFM ANALYSIS


(Recency → How recently did the customer purchase?)




(Frequency → How often did the customer purchase?)


(Monetary → How much did the customer spend?)

```python
snapshot_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'Revenue': 'sum'
}).reset_index()

rfm.columns = [
    'CustomerID',
    'Recency',
    'Frequency',
    'Monetary'
]

rfm.head()
```

CREATE SCORES

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

THEN

```python
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

rfm.head()
```

RFM CUSTOMER SEGMENTATION

```python
def rfm_segment(row):
    if row['R_Score'] >= 4 and row['F_Score'] >= 4 and row['M_Score'] >= 4:
        return 'Champions'
    elif row['R_Score'] >= 3 and row['F_Score'] >= 3:
        return 'Loyal Customers'
    elif row['R_Score'] <= 2 and row['M_Score'] >= 4:
        return 'At Risk'
    elif row['R_Score'] >= 4 and row['F_Score'] <= 2:
        return 'New Customers'
    else:
        return 'Regular Customers'

rfm['Segment'] = rfm.apply(rfm_segment, axis=1)

rfm['Segment'].value_counts()
```

```python
rfm.groupby('Segment').agg(
    Customers=('CustomerID', 'count'),
    Revenue=('Monetary', 'sum'),
    AvgRevenue=('Monetary', 'mean')
).sort_values('Revenue', ascending=False)
```

 CUSTOMER CHURN ANALYSIS

```python
# Define the analysis end date
analysis_end_date = df_clean['InvoiceDate'].max()

# Calculate days since customer's last purchase
customer_summary['DaysSinceLastPurchase'] = (
    analysis_end_date - customer_summary['LastPurchase']
).dt.days

# Define churn status
customer_summary['ChurnStatus'] = customer_summary[
    'DaysSinceLastPurchase'
].apply(
    lambda x: 'Churned' if x > 90 else 'Active'
)

customer_summary['ChurnStatus'].value_counts()
```

CHURN RATE

```python
churn_summary = customer_summary['ChurnStatus'].value_counts()

churn_rate = (
    (customer_summary['ChurnStatus'] == 'Churned').mean() * 100
)

print(f"Customer Churn Rate: {churn_rate:.2f}%")
```

IDENTIFY HIGH VALUE CHURN CUSTOMER

```python
high_value_churned = (
    customer_summary[
        (customer_summary['ChurnStatus'] == 'Churned') &
        (customer_summary['CustomerSegment'] == 'High Value')
    ]
    .sort_values('TotalRevenue', ascending=False)
)

high_value_churned.head(20)
```

CHURN VISUALIZATION

```python
import matplotlib.pyplot as plt

churn_counts = customer_summary['ChurnStatus'].value_counts()

plt.figure(figsize=(7, 5))

churn_counts.plot(kind='bar')

plt.title('Active vs Churned Customers')
plt.xlabel('Customer Status')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
